```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# ===============================
# 1️⃣ Load Dataset
# ===============================
# Make sure data6.txt is in same folder as this script
df = pd.read_csv("data6.txt", sep=r"\s+", header=None)

# Rename first column as Index
df = df.rename(columns={0: "Index"})

# Separate features (exclude index column)
feature_cols = [col for col in df.columns if col != "Index"]
X = df[feature_cols].values

# ===============================
# 2️⃣ Standardize Features
# ===============================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ===============================
# 3️⃣ Apply K-Means Clustering
# ===============================
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to dataframe
df["Cluster"] = clusters

# ===============================
# 4️⃣ Apply PCA (2 Components)
# ===============================
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# ===============================
# 5️⃣ Plot Clusters
# ===============================
plt.figure()

for cluster in np.unique(clusters):
    plt.scatter(
        X_pca[clusters == cluster, 0],
        X_pca[clusters == cluster, 1],
        label=f"Cluster {cluster}"
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering (PCA Projection)")
plt.legend()
plt.grid(True)
plt.show()

# ===============================
# 6️⃣ Print Results
# ===============================
print("Cluster Assignments:")
print(df)
```

## Clustering Algorithm – Step-by-Step Process

### Step 1: Load the Dataset
The dataset is loaded into the program. The first column (index) is ignored, and only the numerical feature columns are selected for clustering.

### Step 2: Feature Selection
A feature matrix is created using only the numeric feature columns from the dataset.

### Step 3: Standardization
Each feature is standardized by subtracting its mean and dividing by its standard deviation.
This ensures all features are on the same scale so that distance calculations are not biased.

### Step 4: Initialize K-Means
The number of clusters is set to k = 2.
The algorithm initializes two centroids.

### Step 5: Assign Data Points to Clusters
For each data point:
- Compute the Euclidean distance to each centroid.
- Assign the point to the nearest centroid.

### Step 6: Update Centroids
For each cluster:
- Recalculate the centroid as the mean of all data points assigned to that cluster.

### Step 7: Repeat Until Convergence
Repeat the assignment and centroid update steps until:
- Cluster assignments stop changing, or
- The centroids stabilize.

Final cluster labels are then obtained.

### Step 8: Apply PCA for Visualization
Principal Component Analysis (PCA) reduces the dataset from 4 dimensions to 2 dimensions.
This is done only for visualization and does not affect the clustering results.

### Final Workflow
Standardize → Apply K-Means → Obtain Cluster Labels → Apply PCA → Plot Clusters

```python
# =====================================
# Import required libraries
# =====================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris           # built-in iris dataset
from sklearn.preprocessing import StandardScaler # feature scaling
from sklearn.cluster import KMeans               # clustering algorithm
from sklearn.decomposition import PCA            # dimensionality reduction


# =====================================
# 1️⃣ Load the Iris Dataset
# =====================================
# Iris dataset contains 150 samples and 4 features
# Features: sepal length, sepal width, petal length, petal width
# Target: species (setosa, versicolor, virginica)

iris = load_iris()

# Convert dataset to pandas dataframe
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add species labels for reference (not used for clustering)
df["Species"] = iris.target


# =====================================
# 2️⃣ Separate Feature Matrix
# =====================================
# Clustering only uses feature values

X = df[iris.feature_names].values


# =====================================
# 3️⃣ Standardize Features
# =====================================
# Standardization makes features have mean = 0 and variance = 1
# This prevents features with larger values dominating clustering

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# =====================================
# 4️⃣ Apply K-Means Clustering
# =====================================
# Iris dataset naturally has 3 groups

kmeans = KMeans(
    n_clusters=3,     # number of clusters
    random_state=42,  # reproducibility
    n_init=10         # number of centroid initializations
)

clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to dataframe
df["Cluster"] = clusters


# =====================================
# 5️⃣ Apply PCA for Visualization
# =====================================
# PCA reduces 4D data → 2D for plotting

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)


# =====================================
# 6️⃣ Plot Clusters
# =====================================
plt.figure()

for cluster in np.unique(clusters):
    plt.scatter(
        X_pca[clusters == cluster, 0],
        X_pca[clusters == cluster, 1],
        label=f"Cluster {cluster}"
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering on Iris Dataset")
plt.legend()
plt.grid(True)

plt.show()


# =====================================
# 7️⃣ Print Results
# =====================================
print("Cluster Assignments:")
print(df)
```

```python
# =====================================
# Import required libraries
# =====================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_openml
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score


# =====================================
# 1️⃣ Load MNIST dataset
# =====================================
# MNIST has 70,000 handwritten digit images
# Each image is 28x28 = 784 pixel features

mnist = fetch_openml('mnist_784', version=1, as_frame=False)

X = mnist.data
y = mnist.target.astype(int)


# =====================================
# 2️⃣ Take a subset for faster execution
# =====================================
# Full MNIST is large and slow for PCA + KMeans
# Using a subset makes it manageable

X = X[:5000]
y = y[:5000]


# =====================================
# 3️⃣ Standardize the data
# =====================================
# Standardization helps give all features equal importance

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# =====================================
# 4️⃣ Apply PCA before clustering
# =====================================
# Reduce 784 dimensions to a smaller number
# This removes noise and keeps important structure

pca = PCA(n_components=50, random_state=42)
X_pca = pca.fit_transform(X_scaled)


# =====================================
# 5️⃣ Apply K-Means clustering
# =====================================
# We choose 10 clusters because MNIST has digits 0 to 9

kmeans = KMeans(n_clusters=10, random_state=42, n_init=20)
clusters = kmeans.fit_predict(X_pca)


# =====================================
# 6️⃣ Reduce again to 2D only for plotting
# =====================================
# The 50-dimensional PCA output is good for clustering
# But for visualization, reduce it further to 2 dimensions

pca_2d = PCA(n_components=2, random_state=42)
X_vis = pca_2d.fit_transform(X_pca)


# =====================================
# 7️⃣ Plot the clusters
# =====================================
plt.figure(figsize=(8, 6))

for cluster in np.unique(clusters):
    plt.scatter(
        X_vis[clusters == cluster, 0],
        X_vis[clusters == cluster, 1],
        label=f"Cluster {cluster}",
        alpha=0.6,
        s=15
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering on MNIST after PCA")
plt.legend()
plt.grid(True)
plt.show()


# =====================================
# 8️⃣ Evaluate clustering quality
# =====================================
# Since cluster labels do not directly match digit labels,
# accuracy is not the right metric.
# Use clustering metrics like ARI and NMI.

ari = adjusted_rand_score(y, clusters)
nmi = normalized_mutual_info_score(y, clusters)

print("Adjusted Rand Index (ARI):", ari)
print("Normalized Mutual Information (NMI):", nmi)


# =====================================
# 9️⃣ Show sample results
# =====================================
result = pd.DataFrame({
    "Actual Digit": y,
    "Cluster": clusters
})

print(result.head(20))
```

In the original MNIST dataset, each image is represented by 784 raw pixel values ranging from 0 to 255, and these features are used directly for clustering. Because the values are not normalized, some features can dominate the Euclidean distance calculation used by K-means, which can lead to noisy or poorly separated clusters. In the scaled MNIST version, the data is first standardized using a method like StandardScaler, which transforms each feature so it has a mean of 0 and a standard deviation of 1. This ensures that all features contribute equally to the distance calculations, preventing large-value features from overpowering others. As a result, clustering algorithms like K-means can better capture the underlying structure of the data and generally produce more meaningful clusters compared to using the raw MNIST data.

```python
# =========================================================
# K-MEANS + PCA ON IRIS AND MNIST SUBSET
# Better graphical output with digit color-coding 0-9
# =========================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import confusion_matrix, accuracy_score
from scipy.optimize import linear_sum_assignment

# ---------------------------------------------------------
# Helper: remap KMeans cluster labels to true labels
# ---------------------------------------------------------
def remap_clusters_to_labels(y_true, y_pred):
    """
    Remap KMeans cluster IDs to best matching true labels
    using Hungarian algorithm.
    """
    cm = confusion_matrix(y_true, y_pred)
    row_ind, col_ind = linear_sum_assignment(-cm)
    mapping = {col: row for row, col in zip(row_ind, col_ind)}
    y_remapped = np.array([mapping[label] for label in y_pred])
    return y_remapped, mapping, cm

# ---------------------------------------------------------
# Helper: plot scatter with clean legend
# ---------------------------------------------------------
def plot_scatter(ax, X2d, labels, title, cmap='tab10', n_classes=None, centers=None):
    scatter = ax.scatter(
        X2d[:, 0], X2d[:, 1],
        c=labels,
        cmap=cmap,
        s=35,
        alpha=0.75,
        edgecolors='k',
        linewidths=0.3
    )

    if centers is not None:
        ax.scatter(
            centers[:, 0], centers[:, 1],
            marker='X',
            s=260,
            c='black',
            edgecolors='white',
            linewidths=1.5,
            label='Centroids'
        )

    ax.set_title(title, fontsize=13, weight='bold')
    ax.set_xlabel("Principal Component 1")
    ax.set_ylabel("Principal Component 2")
    ax.grid(True, alpha=0.25)

    if n_classes is not None:
        handles = []
        cmap_obj = plt.get_cmap(cmap)
        for i in range(n_classes):
            handles.append(
                Line2D(
                    [0], [0],
                    marker='o',
                    color='w',
                    label=str(i),
                    markerfacecolor=cmap_obj(i / max(n_classes - 1, 1)),
                    markeredgecolor='k',
                    markersize=8
                )
            )
        if centers is not None:
            handles.append(
                Line2D(
                    [0], [0],
                    marker='X',
                    color='w',
                    label='Centroids',
                    markerfacecolor='black',
                    markeredgecolor='white',
                    markersize=11
                )
            )
        ax.legend(handles=handles, title="Classes", bbox_to_anchor=(1.02, 1), loc='upper left')

# =========================================================
# PART 1: IRIS
# =========================================================

# Load Iris
iris = load_iris()
X_iris = iris.data
y_iris = iris.target
iris_names = iris.target_names

# Standardize
scaler_iris = StandardScaler()
X_iris_scaled = scaler_iris.fit_transform(X_iris)

# KMeans
kmeans_iris = KMeans(n_clusters=3, random_state=42, n_init=20)
iris_clusters = kmeans_iris.fit_predict(X_iris_scaled)

# Remap clusters to true labels for readability
iris_clusters_remap, iris_mapping, iris_cm = remap_clusters_to_labels(y_iris, iris_clusters)

# PCA
pca_iris = PCA(n_components=2, random_state=42)
X_iris_pca = pca_iris.fit_transform(X_iris_scaled)
iris_centers_pca = pca_iris.transform(kmeans_iris.cluster_centers_)

# Accuracy after remapping
iris_acc = accuracy_score(y_iris, iris_clusters_remap)

# Plot Iris
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

plot_scatter(
    axes[0], X_iris_pca, y_iris,
    title="Iris: True Species (PCA Projection)",
    cmap='viridis',
    n_classes=3
)

plot_scatter(
    axes[1], X_iris_pca, iris_clusters_remap,
    title=f"Iris: K-Means Clusters (Remapped), Accuracy = {iris_acc:.3f}",
    cmap='viridis',
    n_classes=3,
    centers=iris_centers_pca
)

plt.suptitle("K-Means + PCA on Iris Dataset", fontsize=16, weight='bold')
plt.tight_layout()
plt.show()

print("Iris cluster-to-label mapping:", iris_mapping)
print("Iris accuracy after remapping:", round(iris_acc, 4))
print("Iris explained variance ratio:", pca_iris.explained_variance_ratio_)

# =========================================================
# PART 2: MNIST SUBSET
# =========================================================
# In Colab, this should work directly.
# We take a balanced subset for cleaner clustering and plotting.

from tensorflow.keras.datasets import mnist

# Load MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Combine train + test
X_mnist = np.concatenate([X_train, X_test], axis=0)
y_mnist = np.concatenate([y_train, y_test], axis=0)

# Flatten images: 28x28 -> 784
X_mnist = X_mnist.reshape(len(X_mnist), -1).astype(np.float32)

# ---------------------------------------------------------
# Build a balanced subset: same number per digit 0-9
# ---------------------------------------------------------
samples_per_digit = 300   # adjust if you want more
subset_indices = []

for digit in range(10):
    digit_idx = np.where(y_mnist == digit)[0][:samples_per_digit]
    subset_indices.extend(digit_idx)

subset_indices = np.array(subset_indices)
X_mnist_sub = X_mnist[subset_indices]
y_mnist_sub = y_mnist[subset_indices]

print("\nMNIST subset shape:", X_mnist_sub.shape)
print("MNIST subset labels shape:", y_mnist_sub.shape)

# Standardize
scaler_mnist = StandardScaler()
X_mnist_scaled = scaler_mnist.fit_transform(X_mnist_sub)

# KMeans with 10 clusters
kmeans_mnist = KMeans(n_clusters=10, random_state=42, n_init=20)
mnist_clusters = kmeans_mnist.fit_predict(X_mnist_scaled)

# Remap clusters to digit labels 0-9
mnist_clusters_remap, mnist_mapping, mnist_cm = remap_clusters_to_labels(y_mnist_sub, mnist_clusters)
mnist_acc = accuracy_score(y_mnist_sub, mnist_clusters_remap)

# PCA to 2D
pca_mnist = PCA(n_components=2, random_state=42)
X_mnist_pca = pca_mnist.fit_transform(X_mnist_scaled)
mnist_centers_pca = pca_mnist.transform(kmeans_mnist.cluster_centers_)

# Plot MNIST
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

plot_scatter(
    axes[0], X_mnist_pca, y_mnist_sub,
    title="MNIST Subset: True Digits (0–9), PCA Projection",
    cmap='tab10',
    n_classes=10
)

plot_scatter(
    axes[1], X_mnist_pca, mnist_clusters_remap,
    title=f"MNIST Subset: K-Means Clusters (Remapped), Accuracy = {mnist_acc:.3f}",
    cmap='tab10',
    n_classes=10,
    centers=mnist_centers_pca
)

plt.suptitle("K-Means + PCA on MNIST Subset", fontsize=16, weight='bold')
plt.tight_layout()
plt.show()

print("MNIST cluster-to-digit mapping:", mnist_mapping)
print("MNIST accuracy after remapping:", round(mnist_acc, 4))
print("MNIST explained variance ratio:", pca_mnist.explained_variance_ratio_)

# =========================================================
# OPTIONAL: Show confusion matrices
# =========================================================
print("\nIris confusion matrix before remapping:")
print(iris_cm)

print("\nMNIST confusion matrix before remapping:")
print(mnist_cm)
```
